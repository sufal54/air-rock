# Generic Netlink Message Structure

A **Generic Netlink message is layered**. Each layer has its own header and purpose.

```text
+---------------------+
|      nlmsghdr       |  <-- Netlink header
+---------------------+
|     genlmsghdr      |  <-- Generic Netlink header
+---------------------+
|       nlattr        |  <-- Attribute
+---------------------+
|       nlattr        |
+---------------------+
|         ...         |
```

---

## 1. `nlmsghdr` — Netlink Header

This header is understood by the **Linux Netlink subsystem itself**.

### Fields

* `nlmsg_len` → Total message length
* `nlmsg_type` → Message type (family ID or kernel message type)
* `nlmsg_flags` → Request flags
* `nlmsg_seq` → Sequence number
* `nlmsg_pid` → Port ID (usually your socket ID)

### Common Flags

```text
NLM_F_REQUEST       = 0x0001
NLM_F_MULTI         = 0x0002
NLM_F_ACK           = 0x0004
NLM_F_ECHO          = 0x0008
NLM_F_DUMP_INTR     = 0x0010
NLM_F_DUMP_FILTERED = 0x0020

NLM_F_ROOT          = 0x0100
NLM_F_MATCH         = 0x0200
NLM_F_ATOMIC        = 0x0400
NLM_F_DUMP          = 0x0300

NLM_F_REPLACE       = 0x0100
NLM_F_EXCL          = 0x0200
NLM_F_CREATE        = 0x0400
NLM_F_APPEND        = 0x0800

NLM_F_NONREC        = 0x0100
NLM_F_BULK          = 0x0200

NLM_F_CAPPED        = 0x0100
NLM_F_ACK_TLVS      = 0x0200
```

### Special `nlmsg_type` Values

```text
NLMSG_NOOP       = 0x1
NLMSG_ERROR      = 0x2
NLMSG_DONE       = 0x3
NLMSG_OVERRUN    = 0x4

GENL_ID_CTRL = 0x10
```

Otherwise, `nlmsg_type` is usually the **Generic Netlink family ID**.

For example:

```text
nl80211
```

> **Important:** `nlmsg_type` identifies the Generic Netlink family, while `genlmsghdr.cmd` identifies the command inside that family.

## 2. `genlmsghdr` — Generic Netlink Header

* `cmd` → Which command to execute
* `version` → Protocol version
* `reserved`

# Aliases

These do not have new command IDs:

```text
NL80211_CMD_NEW_BEACON       = NL80211_CMD_START_AP
NL80211_CMD_DEL_BEACON       = NL80211_CMD_STOP_AP

NL80211_CMD_REGISTER_ACTION  = NL80211_CMD_REGISTER_FRAME
NL80211_CMD_ACTION           = NL80211_CMD_FRAME
NL80211_CMD_ACTION_TX_STATUS = NL80211_CMD_FRAME_TX_STATUS
```

Each **Generic Netlink family defines its own command list**.

For example:

```text
nl80211
    |
    +-- GET_WIPHY
    +-- SET_INTERFACE
    +-- TRIGGER_SCAN
    +-- CONNECT
```

## 3. `nlattr` — Netlink Attributes

Attributes contain the **parameters for the command**.

Each attribute contains:

```text
+------------------+
|     nla_len      |
+------------------+
|     nla_type     |
+------------------+
|      Data        |
+------------------+
```

### Example: `SET_INTERFACE`

Attributes:

```text
IFINDEX = 3
IFTYPE  = MONITOR
```

Conceptually:

```text
SET_INTERFACE(
    IFINDEX = 3,
    IFTYPE  = MONITOR
)
```

### Example: `GET_WIPHY`

```text
WIPHY = 0
SPLIT_WIPHY_DUMP = true
```

They are used to **create, bind, send, and receive Netlink messages**.

The socket API does not understand `nl80211` commands.

It only provides the communication mechanism.

---

## 5. `iovec`

`iovec` tells `sendmsg()` **where each piece of data is located**.

Conceptually:

```text
iov[0] -> nlmsghdr
iov[1] -> genlmsghdr
iov[2] -> attribute1
iov[3] -> attribute2
```

Example:

```text
+------------------+
| iov[0]           | ---> nlmsghdr
+------------------+
| iov[1]           | ---> genlmsghdr
+------------------+
| iov[2]           | ---> nlattr
+------------------+
| iov[3]           | ---> nlattr
+------------------+
```

`iovec` is basically an **array of memory buffers**.

---

## 6. `msghdr`

`msghdr` combines everything needed by `sendmsg()` or `recvmsg()`.

It contains:

```text
sockaddr_nl
iovec array
control data
flags
```

Conceptually:

```text
msghdr
  |
  +-- sockaddr_nl
  |
  +-- iovec array
  |     |
  |     +-- nlmsghdr
  |     +-- genlmsghdr
  |     +-- nlattr
  |     +-- nlattr
  |
  +-- control data
  |
  +-- flags
```

It is passed to:

```c
sendmsg()
```

or:

```c
recvmsg()
```

---

# Complete Structure Summary

| Layer          | Structure    | Purpose                          |
| -------------- | ------------ | -------------------------------- |
| Socket         | `socket()`   | Create Netlink socket            |
| Message        | `msghdr`     | Container for send/receive       |
| Buffers        | `iovec`      | Points to message parts          |
| Header         | `nlmsghdr`   | Netlink routing, flags, sequence |
| Generic Header | `genlmsghdr` | Command such as GET, SET, NEW    |
| Parameters     | `nlattr`     | Command arguments                |

---

# Think of It Like a Function Call

Imagine this:

```text
socket() -> Open communication

sendmsg(
    nlmsghdr(REQUEST),
    genlmsghdr(SET_INTERFACE),
    attr(IFINDEX = 3),
    attr(IFTYPE = MONITOR)
)
```

The layers mean:

```text
nlmsghdr
    ↓
"This is a Netlink request"

genlmsghdr
    ↓
"Run the SET_INTERFACE command"

nlattr
    ↓
"Use interface 3"

nlattr
    ↓
"Set its type to MONITOR"
```

## Simple Mental Model

```text
Netlink
   ↓
Generic Netlink
   ↓
Family Command
   ↓
Attributes
```

In one sentence:

> **`nlmsghdr` says how Netlink should handle the message, `genlmsghdr` says which Generic Netlink command to run, and `nlattr` provides the command's arguments.**
